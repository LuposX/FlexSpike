```python
import os
import pandas as pd

# datasets folder relative to current working directory
dataset_folder = 'datasets'

normal_datasets = ['Dataset_acuteinflammation.ds',
                   'Dataset_balancescale.ds',
                   'Dataset_breastcancerwisc.ds',
                   'Dataset_cardiotocography3clases.ds',
                   'Dataset_energyy1.ds',
                   'Dataset_energyy2.ds',
                   'Dataset_iris.ds',
                   'Dataset_mammographic.ds',
                   'Dataset_pendigits.ds',
                   'Dataset_seeds.ds',
                   'Dataset_tictactoe.ds',
                   'Dataset_vertebralcolumn2clases.ds',
                   'Dataset_vertebralcolumn3clases.ds']

for ds in normal_datasets:
    path = os.path.join(dataset_folder, ds)
    try:
        data = pd.read_csv(path, sep=',', header=None)  # adjust separator if needed
        X = data.iloc[:, :-1].values
        y = data.iloc[:, -1].values
        n_features = X.shape[1]
        n_classes = len(pd.unique(y))
        print(f"{ds}: features={n_features}, classes={n_classes}")
    except Exception as e:
        print(f"Could not load {ds}: {e}")
```

```python
import os
from utils.Loader import GetDataLoader

# assume args object exists with all required attributes
# if running standalone, we can mock it:
class Args:
    DataPath = './dataset'  # folder with your datasets
    task = 'normal'          # placeholder
    DATASET = 0              # will be overwritten in loop
    augment = False
    NOISE_LEVEL = 0
    WARP_FACTOR = 1.0
    SFR_down = 1.0
    SFR_up = 1.0
    CROP_SIZE = None
    R_train = 20
    R_valid = 10
    R_test = 5
    DEVICE = "cpu"


args = Args()

# Dataset lists from your function
normal_datasets = ['Dataset_acuteinflammation.ds',
                   'Dataset_balancescale.ds',
                   'Dataset_breastcancerwisc.ds',
                   'Dataset_cardiotocography3clases.ds',
                   'Dataset_energyy1.ds',
                   'Dataset_energyy2.ds',
                   'Dataset_iris.ds',
                   'Dataset_mammographic.ds',
                   'Dataset_pendigits.ds',
                   'Dataset_seeds.ds',
                   'Dataset_tictactoe.ds',
                   'Dataset_vertebralcolumn2clases.ds',
                   'Dataset_vertebralcolumn3clases.ds']

temporal_datasets = ['Dataset_cbf.tsds',
                     'Dataset_distalphalanxtw.tsds',
                     'Dataset_freezerregulartrain.tsds',
                     'Dataset_freezersmalltrain.tsds',
                     'Dataset_gunpointagespan.tsds',
                     'Dataset_gunpointmaleversusfemale.tsds',
                     'Dataset_gunpointoldversusyoung.tsds',
                     'Dataset_middlephalanxoutlineagegroup.tsds',
                     'Dataset_mixedshapesregulartrain.tsds',
                     'Dataset_powercons.tsds',
                     'Dataset_proximalphalanxoutlinecorrect.tsds',
                     'Dataset_selfregulationscp2.tsds',
                     'Dataset_slope.tsds',
                     'Dataset_smoothsubspace.tsds',
                     'Dataset_symbols.tsds']

# -----------------------------
# Function to extract features/classes using GetDataLoader
# -----------------------------
def extract_info(dataset_list, task_type):
    results = []
    args.task = task_type
    for i, ds_name in enumerate(dataset_list):
        args.DATASET = i
        # load train loader and info (we only need info)
        try:
            _, info = GetDataLoader(args, mode='train', path=args.DataPath)
            results.append((ds_name, info['N_feature'], info['N_class']))
        except Exception as e:
            results.append((ds_name, 'ERROR', 'ERROR'))
            print(f"Error loading {ds_name}: {e}")
    return results

# -----------------------------
# Get info for all datasets
# -----------------------------
all_info = []
all_info += extract_info(normal_datasets, 'normal')
all_info += extract_info(temporal_datasets, 'temporal')

# -----------------------------
# Print table
# -----------------------------
print(f"{'Dataset':40s} | {'Features':8s} | {'Classes':8s}")
print("-"*65)
for ds_name, n_feat, n_class in all_info:
    print(f"{ds_name:40s} | {str(n_feat):8s} | {str(n_class):8s}")
```
